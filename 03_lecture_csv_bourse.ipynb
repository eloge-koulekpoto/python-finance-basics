# Lecture d'un historique boursier au format CSV

Lecture de l'historique quotidien d'Apple téléchargé sur stooq.com
(10 584 séances depuis 1984) et calcul, sur les séances depuis 2024 :
prix moyen de clôture, plus haut et plus bas avec leurs dates.

Deux points traités ici : le module `csv` plutôt qu'un découpage manuel,
qui gère les virgules à l'intérieur d'un champ, et la gestion des données
imparfaites — lignes vides et cotations manquantes sont écartées et
comptées au lieu d'interrompre le programme.

**Données** : `data/aapl_us_d.csv` — source stooq.com, format
`Date,Open,High,Low,Close,Volume`.

```python
import csv
count = 0
count_float_echec = 0
close_list = []
date_list = []

try:
    with open("data/aapl_us_d.csv",encoding="utf-8") as fhand:

        # csv.reader splits each line into a list of fields
        lecteur = csv.reader(fhand)

        # next() consumes the header row so the loop starts at the data
        print("en-tête:",next(lecteur))

        for liste in lecteur:
            if liste == []:
                continue

            # ISO dates compare as plain strings, no conversion needed   
            if liste[0] >= "2024-01-01":
                
                try:
                    x = float(liste[4])
                    close_list.append(x)
                    date_list.append(liste[0])
                    count = count + 1
        
                except ValueError:
                    # missing quotes are counted, not fatal
                    count_float_echec = count_float_echec +1
                
    print(f"Période étudiée: du {date_list[0]} au {date_list[-1]}")
    print(f"le prix moyen: {sum(close_list)/len(close_list):.2f}")
    print(f"le plus haut: {max(close_list):.2f} le {date_list[close_list.index(max(close_list))]};le plus bas: {min(close_list):.2f} le {date_list[close_list.index(min(close_list))]}")
    print(f"Nombre de cotations manquantes: {count_float_echec}")
    
except FileNotFoundError :
    print("Can't open file")
```
