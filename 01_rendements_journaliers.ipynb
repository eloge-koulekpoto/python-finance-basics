# Rendements journaliers d'un titre

Lecture d'un fichier au format `date;prix` et calcul du rendement quotidien
d'une séance à l'autre : rendement moyen sur la période, meilleure et pire
séance avec leurs dates.

Le fichier est parcouru une seule fois, sans stocker les prix : le prix de
la veille est conservé dans une variable réutilisée à chaque tour.

**Données** : `data/prix.txt` — 10 séances (août-septembre 2026).

```python
with open("data/prix.txt") as fhand:
    
    # no previous close on the first line: None marks that case
    closeprice1 = None
    count = 0
    tot = 0
    smallest_yield = None
    highest_yield = None
    sdate = None
    hdate = None

    for line in fhand:

        # anchor on the separator, not on a digit position
        x = line.find(";")
        closeprice2 = float(line[x+1:])
        if closeprice1 is not None:
            daily_yield = (closeprice2 - closeprice1) / closeprice1
            count = count + 1
            tot = tot + daily_yield

            # None as starting point: any real return replaces it, even a negative one
            if smallest_yield is None or daily_yield < smallest_yield:
                smallest_yield = daily_yield
                sdate = line[:x]
            if highest_yield is None or daily_yield > highest_yield:
                highest_yield = daily_yield
                hdate = line[:x]

        # outside the if: today's close becomes tomorrow's previous close
        closeprice1 = closeprice2

print(f"The average yield is {tot/count:+.2%}; The worst day is {sdate} with {smallest_yield:+.2%}; The best day is {hdate} with {highest_yield:+.2%}")
```
