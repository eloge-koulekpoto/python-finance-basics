# Rendements par titre

Lecture d'un fichier de cotations multi-titres au format `ticker;date;prix`
et calcul, pour chaque titre, du rendement moyen quotidien et du rendement
cumulé sur la période.

Les tickers ne sont écrits nulle part dans le code : ils sont découverts
à la lecture du fichier. Ajouter un titre au fichier ne demande aucune
modification du programme.

**Données** : `data/cotations.txt` — 4 titres, 10 séances (août-septembre 2026).

```python
# maps each ticker to the list of its closing prices
prix = {}

with open("data/cotations.txt", encoding="utf-8") as fhand:
    for line in fhand:
        line = line.strip()
        
        # skip blank lines: split() would fail on them
        if line == "":
            continue
            
        ticker, date, valeur = line.split(";")

        # setdefault creates the list on first encounter with this ticker
        prix.setdefault(ticker, []).append(float(valeur))

for ticker, valeurs in prix.items():
    rendements = []
    for i in range(1, len(valeurs)):
        rendements.append((valeurs[i] - valeurs[i-1]) / valeurs[i-1])

    moyenne = sum(rendements) / len(rendements)

    # compounded, not the sum of daily returns
    cumule = (valeurs[-1] - valeurs[0]) / valeurs[0]

    print(f"{ticker:5} moyenne {moyenne:+.2%}   cumulé {cumule:+.2%}   {len(rendements)} séances")
```
